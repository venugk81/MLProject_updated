## PIL - Python Imaging Library

- Python library used for working with images.
- Nowadays the updated version of PIL is used.
- Pillow is basically the modern version of PIL.

#### Installation Command - pip install pillow

```python
from PIL import Image
```

```python
# Opening the image
img = Image.open("cat.jpg")
```

```python
# Displaying the image here
img
```

```python
# Displaying an image
img.show()
```

```python
print(img.size)
print(img.format)
print(img.mode)
```

```python
# Resizing image - changing dimensions
resized = img.resize((300,300))
resized
```

```python
# resized.show()
```

```python
 # Rotating an image
rotated = resized.rotate(180)
```

```python
rotated
```

```python
# Cropping an image
cropped = resized.crop((100, 100, 400, 400))    # (left, top, right, bottom)
```

```python
cropped
```

```python
# cropped.show()
```

```python
# Saving a modified Image
rotated.save('Rotated_cat.jpg')
```

```python
# Converting one image format to another
resized.save('cat.png')
```

```python
# Converting the scale of the image
cat = Image.open('cat.png')
```

```python
cat.show()
```

```python
cat_L = cat.convert('L')
```

```python
cat_L
```

```python
cat_L.mode
```

mode = L is a 8-bit grayscale image

#### Enhancing Image

```python
from PIL import ImageEnhance
```

```python
rose = Image.open('rose.jpg')
```

```python
rose
```

```python
enhancer = ImageEnhance.Brightness(rose)
enhancer
```

```python
rose2 = enhancer.enhance(0.5)
rose2
```

```python
rose3 = enhancer.enhance(2)
rose3
```

#### Filtering the image

```python
from PIL import ImageFilter
```

```python
cat = Image.open('cat.png')
cat
```

```python
# Blurring the image
blur = cat.filter(ImageFilter.BLUR)
```

```python
blur
```

```python
# Flipping the image
flip = cat.transpose(Image.FLIP_LEFT_RIGHT)
flip
```

```python
# Flipping the image
flip2 = cat.transpose(Image.FLIP_TOP_BOTTOM)
flip2
```
