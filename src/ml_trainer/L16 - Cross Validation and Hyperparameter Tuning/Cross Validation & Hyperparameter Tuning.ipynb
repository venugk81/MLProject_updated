```python
import pandas as pd
```

```python
df = pd.read_csv('multiclass.csv')
```

```python
X = df.drop('class', axis=1)
y = df['class']
```

#### Data Splitting

```python
from sklearn.model_selection import train_test_split
```

```python
X_train_val, X_test, y_train_val, y_test = train_test_split(X, y, test_size=0.15, random_state=0)
```

```python
X_train, X_val, y_train, y_val = train_test_split(X_train_val, y_train_val, test_size=0.15, random_state=0)
```

#### Scaling the data

```python
X_train.head(3)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_train[:3]
```

```python
X_val = scaler.transform(X_val)
```

```python
X_val[0]
```

```python
X_test = scaler.transform(X_test)
```

```python
X_test[0]
```

- Dividing the data into train, validation and test sets is different from cross validation.

#### Model Building

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn1 = KNeighborsClassifier()
```

```python
knn1.fit(X_train, y_train)
```

```python
knn1.score(X_train, y_train)
```

```python
knn1.score(X_val, y_val)
```

#### Model Building using Cross Validation

```python
from sklearn.model_selection import cross_val_score
```

```python
knn2 = KNeighborsClassifier()
```

```python
cross_validate = cross_val_score(knn2, X_train, y_train, cv=5, scoring='accuracy', n_jobs=-1)
```

```python
cross_validate
```

```python
cross_validate.mean()
```

#### K-Fold Cross Validation Method

```python
from sklearn.model_selection import KFold
```

```python
knn3 = KNeighborsClassifier()
```

```python
kf = KFold(n_splits=10, shuffle=True, random_state=0)
```

```python
cross_validate = cross_val_score(knn3, X_train, y_train, cv=kf, scoring='accuracy', n_jobs=-1)
```

```python
cross_validate
```

```python
cross_validate.mean()
```

Training the final model

```python
knn3.fit(X_train, y_train)
```

```python
knn3.score(X_train, y_train)
```

```python
knn3.score(X_val, y_val)
```

```python
knn3.score(X_test, y_test)
```

#### Hyperparameter Tuning Techniques
1. GridSearchCV
2. RandomisedSearchCV

##### GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
```

```python
knn = KNeighborsClassifier()
```

```python
# Defining hyperparameters
param_grid = {
    'n_neighbors': [3, 5, 7, 9],
    'p': [1, 2, 3, 4],
    'weights': ['uniform', 'distance'],
}
```

```python
grid_search = GridSearchCV(estimator=knn, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1)
```

```python
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
grid_search.best_score_
```

```python
# grid_search.best_estimator_
```

```python
best_knn = grid_search.best_estimator_
```

```python
best_knn.score(X_train, y_train)
```

```python
best_knn.score(X_val, y_val)
```

```python
best_knn.score(X_test, y_test)
```

##### RandomisedSearchCV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
knn = KNeighborsClassifier()
```

```python
# Defining hyperparameters
param_grid = {
    'n_neighbors': [3, 5, 7, 9],
    'p': [1, 2, 3, 4],
    'weights': ['uniform', 'distance'],
}
```

```python
random_search = RandomizedSearchCV(estimator=knn, param_distributions=param_grid, n_iter=20, cv=5, scoring='accuracy',
                                   random_state=42, n_jobs=-1)
```

```python
random_search.fit(X_train, y_train)
```

```python
random_search.best_params_
```

```python
random_search.best_score_
```

```python
best_knn = random_search.best_estimator_
```

```python
best_knn.score(X_train, y_train)
```

```python
best_knn.score(X_val, y_val)
```

```python
best_knn.score(X_test, y_test)
```
