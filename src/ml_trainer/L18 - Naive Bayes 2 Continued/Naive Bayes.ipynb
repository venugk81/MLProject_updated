## Email Classification using Naive Bayes

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
```

```python
df = pd.read_csv('email_dataset.csv', index_col=0)
df.head()
```

```python
df.info()
```

#### Encoding the Columns

```python
df['Label'] = df['Label'].apply(lambda x:0 if x=='ham' else 1)
```

```python
df['Label'].head()
```

#### Splitting the dataset

```python
X = df['Email']
y = df['Label']
```

```python
X.head()
```

```python
y.head()
```

#### Divding the data into train and test sets

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
X_train.shape, y_train.shape
```

```python
X_test.shape, y_test.shape
```

#### Encoding the Email column

```python
tfidf = TfidfVectorizer()
```

```python
X_train = tfidf.fit_transform(X_train)
X_test = tfidf.transform(X_test)
```

```python
X_train, X_test
```

- Total unique words present in the email dataset is 7612
- 7612 are the number of columns created after encoding the text
- Each word acts as a feature/column in the new dataset
- For a specific email, if a word is present, the value asigned to that respective column is 1 else 0

```python
X_train = X_train.toarray()
X_train
```

```python
X_test = X_test.toarray()
X_test
```

#### Model Building

```python
model = MultinomialNB()
```

```python
model.fit(X_train, y_train)
```

```python
print(f'Training Accuracy: {model.score(X_train, y_train)}')
```

```python
print(f'Test Accuracy: {model.score(X_test, y_test)}')
```

#### Prediction on unseen data

```python
new_email = ['Good morning, are you coming today?']
```

```python
new_email = tfidf.transform(new_email)
```

```python
new_email
```

```python
new_email.toarray()
```

```python
prediction = model.predict(new_email)
prediction
```

```python
if prediction[0] == 1:
    print("Spam")
else:
    print("Ham")
```
