#### Groupby() 
* it is used to make groups of similar data and then perform calulations on each gtoup
* it works in 3 simple steps
* 1- split ---> Divide the data into group
* 2- Apply ---> Perform some operation on each group
* 3- combine --> the final result together

### Pivot Table
* it is a tool for summerizing and analyzing data.
* 1- pivot() and pivot_table() ---> used to convert rows into columns
* 2- stack() and unstack() ---> used to switch rows and columns, -- stack()=columns become rows, unstack() = rows become column
* 3- melt() ----> convert wide data into long data
* 4- get_dummies() --->  convert text categories into numbers(0 and 1)
* 5- explode() --->Splits list values into separate rows
* 6- cut() ----> convert continous numbers into categories --(0-30), (30-90),(90-110)
* 7--factorize() ---> Convert text labels into integer numbers 

```python
# lets create a Data Frame
dict1 = {"seq":['one','one','one','two','two','two'],
         "alp":['A','B','C','D','E','F'],
         "num":[1,2,3,1,2,3],}
df = pd.DataFrame(dict1)
print(df)
```

```python
# lets use pivot()
df.pivot(index='seq', columns='alp', values='num')
```

```python
#lets do one more time 
df.pivot(index='alp', columns='seq', values='num')
```

```python
# another example
dict1 = {"seq":['one','one','one','two','two','two'],
         "alp":['A','B','C','D','E','F'],
         "num":[1,2,3,1,2,"NAN"],}
df2 = pd.DataFrame(dict1)
print(df2)
```

```python
# sample Dataframe 
data = {
    'Region':['North','South','North','South','North','South'],
    'Product':['A','A','B','B','C','C',],
    'Sales':[100,200,150,300,250, 400]
}

df = pd.DataFrame(data)
df
```

```python
# lets use pivot table
pivot = pd.pivot_table(df,
                       values = 'Sales',
                       index = 'Region',
                       columns = 'Product',
                       aggfunc = 'sum', # operation addition is to be perform
                       fill_value=0, # if there is NaN , will replace this with 0
                       margins = True) # Total value of operation
pivot
```

### Difference between Groupby and pivot table 

*  **Groupby** : this is used to group data and apply aggregation or transformation functions to each group.
   its more general and flexible for operations like Summing, averaging and filtering data groups
*  **Pivot Table** : it is specifically designed to create pivot table for summarization.
   it allows reshaping data into tabular format with rows, columns and aggregated values .
